```python
# choosing method of imputation
# mean, median mode and KNNImputer
```

```python
df.isnull().sum()
```

# **1) Numerical Data**

```python
# using Mean and median method to handle missing values
# to handle numerical data
for i in ["BMI", "Polio", "Income composition of resources"]:
    df[i].fillna(df[i].median(), inplace=True)
```

```python
df.isnull().sum()
```

# **2) To handle both numerical and categorical**

```python
from sklearn.impute import KNNImputer

impute = KNNImputer()
```

```python
for i in df.select_dtypes(include="number").columns:
    df[i] = impute.fit_transform(df[[i]])
```
